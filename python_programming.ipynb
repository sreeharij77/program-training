```python
n=int(input('number of monkeys'))
k=int(input('number of eatable bananas'))
j=int(input('number of eatable peanut'))
m=int(input('number of bananas'))
p=int(input('number of peanuts'))
if n>0 or p>0 or m>0 or k>0 or j>0:
  n=n-((m//k)+(p//j))
  if m%k!=0 and p%j!=0:
    n=n-1
else:
  print("invlid")
print("remaining monkeys",n)

```
